```python
from transformers import BertTokenizer, BertModel
import torch

# 1) Load tokenizer and model
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')

# 2) Tokenize text
text = "This is a sample sentence."
encoded = tokenizer(text,
                    padding=True,
                    truncation=True,
                    return_tensors="pt")  # returns PyTorch tensors
input_ids = encoded['input_ids']
attention_mask = encoded['attention_mask']

# 3) Get model output (embeddings)
with torch.no_grad():  # turn off gradients (we're not training)
    output = model(input_ids=input_ids,
                   attention_mask=attention_mask)

# 4) Extract embedding(s)
# output.last_hidden_state has shape [batch_size, sequence_length, hidden_size]
token_embeddings = output.last_hidden_state
# If you want one embedding representing the full sentence, you might pick the [CLS] token embedding:
sentence_embedding = token_embeddings[:, 0, :]  # shape [batch_size, hidden_size]
```
